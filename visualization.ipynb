```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("ggplot")

pd.set_option("display.max_columns", None)
```

```python
df = pd.read_csv("/content/dataset.csv")

df.head()
```

```python
plt.figure(figsize=(10,5))

sns.heatmap(df.isnull(), cbar=False)

plt.title("Missing Values Heatmap")

plt.show()
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
numerical = df.select_dtypes(include=np.number)

categorical = df.select_dtypes(exclude=np.number)

print(numerical.columns)

print(categorical.columns)
```

```python
plt.figure(figsize=(12,6))

genre_counts = df['track_genre'].value_counts().head(15)

genre_counts.plot(kind='bar', color='skyblue')

plt.title("Top 15 Spotify Genres")
plt.xlabel("Genre")
plt.ylabel("Number of Tracks")
plt.xticks(rotation=45)

plt.savefig("images/genre_distribution.png",
            dpi=300,
            bbox_inches="tight")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.histplot(df['popularity'],
             bins=30,
             kde=True)

plt.title("Distribution of Song Popularity")
plt.xlabel("Popularity")
plt.ylabel("Count")

plt.savefig("images/popularity_distribution.png",
            dpi=300,
            bbox_inches="tight")

plt.show()
```

```python
plt.figure(figsize=(6,5))

sns.countplot(data=df,
              x='explicit')

plt.title("Explicit vs Non-Explicit Songs")
plt.xlabel("Explicit")
plt.ylabel("Count")

plt.savefig("images/explicit_songs.png",
            dpi=300,
            bbox_inches="tight")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    data=df,
    x='danceability',
    y='popularity',
    alpha=0.6
)

plt.title("Danceability vs Popularity")
plt.xlabel("Danceability")
plt.ylabel("Popularity")

plt.savefig("images/danceability_vs_popularity.png",
            dpi=300,
            bbox_inches="tight")

plt.show()
```

```python
plt.figure(figsize=(10,4))

sns.boxplot(x=df['energy'])

plt.title("Distribution of Energy")

plt.savefig("images/energy_boxplot.png",
            dpi=300,
            bbox_inches="tight")

plt.show()
```

```python
numeric_df = df.select_dtypes(include='number')

plt.figure(figsize=(14,10))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap='coolwarm',
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.savefig("images/correlation_heatmap.png",
            dpi=300,
            bbox_inches="tight")

plt.show()
```

```python
plt.figure(figsize=(12,6))

artist_counts = df['artists'].value_counts().head(15)

artist_counts.plot(kind='bar',
                   color='orange')

plt.title("Top 15 Artists")
plt.xlabel("Artist")
plt.ylabel("Number of Songs")
plt.xticks(rotation=60)

plt.savefig("images/top_artists.png",
            dpi=300,
            bbox_inches="tight")

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.histplot(df['loudness'],
             bins=35,
             kde=True)

plt.title("Distribution of Loudness")
plt.xlabel("Loudness")
plt.ylabel("Count")

plt.savefig("images/loudness_distribution.png",
            dpi=300,
            bbox_inches="tight")

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.histplot(df['tempo'],
             bins=35,
             kde=True)

plt.title("Distribution of Tempo")
plt.xlabel("Tempo")
plt.ylabel("Count")

plt.savefig("images/tempo_distribution.png",
            dpi=300,
            bbox_inches="tight")

plt.show()
```

```python
genre_popularity = (
    df.groupby("track_genre")["popularity"]
      .mean()
      .sort_values(ascending=False)
      .head(15)
)

plt.figure(figsize=(12,6))

genre_popularity.plot(kind='bar',
                      color='green')

plt.title("Top Genres by Average Popularity")
plt.xlabel("Genre")
plt.ylabel("Average Popularity")
plt.xticks(rotation=45)

plt.savefig("images/popularity_by_genre.png",
            dpi=300,
            bbox_inches="tight")

plt.show()
```
